```python
# Importações para aplicar transformações necessárias
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
from sklearn.model_selection import train_test_split
```

```python
df_taxi = pd.read_csv('datasets/taxi_tripe_pricing.csv')
```

```python
# Como feito no EDA iremos remover todas as linhas com dados nulos para simplificar o dataset
df_taxi.isnull().sum()
```

```python
# Remove as linhas com valores nulos
df_taxi = df_taxi.dropna()
```

```python
# Separando variáveis preditoras e alvo
X = df_taxi.drop("Trip_Price", axis=1)
y = df_taxi["Trip_Price"]

# Separação em treino e teste (30% para teste)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Exporta os datasets em CSV
train_df = pd.concat([X_train, y_train], axis=1)
test_df = pd.concat([X_test, y_test], axis=1)

train_df.to_csv("datasets/taxi_tripe_pricing_train.csv", index=False)
test_df.to_csv("datasets/taxi_tripe_pricing_test.csv", index=False)
```

```python
# Variáveis numéricas
numericas = [
    "Trip_Distance_km",
    "Passenger_Count",
    "Base_Fare",
    "Per_Km_Rate",
    "Per_Minute_Rate",
    "Trip_Duration_Minutes",
]

# Variáveis categóricas nominais
categoricas_nominais = [
    "Time_of_Day",
    "Day_of_Week",
    "Weather"
]

# Variáveis categóricas ordinais
categoricas_ordinais = [
    "Traffic_Conditions"
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numericas),
        ('cat', OneHotEncoder(), categoricas_nominais),
        ('ord', OrdinalEncoder(), categoricas_ordinais),
    ]
)
```

```python
# Aplicar Column Transformer
preprocessor.fit(train_df)

# Exporta o modelo de pré-processamento
import joblib

joblib.dump(preprocessor, 'pkl/preprocessor_dataset_taxi.pkl')
```
