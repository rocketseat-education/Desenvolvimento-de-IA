```python
# Importações para aplicar transformações necessárias
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, OrdinalEncoder
```

```python
df_airline_train = pd.read_csv('datasets/airline_train.csv')
df_airline_test = pd.read_csv('datasets/airline_test.csv')
```

```python
# Variável Satisfaction será transformada em 0 e 1
df_airline_train['satisfaction'] = df_airline_train['satisfaction'].map({'satisfied': 1, 'neutral or dissatisfied': 0})
df_airline_test['satisfaction'] = df_airline_test['satisfaction'].map({'satisfied': 1, 'neutral or dissatisfied': 0})
```

```python
# Como descoberto no EDA a Coluna Arrival Delay in Minutes possui valores nulos, para solucionar o problema. Vamos adicionar a média da coluna nas linhas faltantes para os 2 datasets
media_arrival_delay = df_airline_train['Arrival Delay in Minutes'].mean()
df_airline_train['Arrival Delay in Minutes'] = df_airline_train['Arrival Delay in Minutes'].fillna(media_arrival_delay)
df_airline_test['Arrival Delay in Minutes'] = df_airline_test['Arrival Delay in Minutes'].fillna(media_arrival_delay)
```

```python
df_airline_train.drop(columns=['Unnamed: 0', 'id'], inplace=True)
df_airline_test.drop(columns=['Unnamed: 0', 'id'], inplace=True)
```

```python
# Salvar dataframe como CSV
df_airline_train.to_csv('./datasets/airline_train_cleaned.csv', index=False)
df_airline_test.to_csv('./datasets/airline_test_cleaned.csv', index=False)
```

```python
# Variáveis categóricas nominais
categoricas_nominais = [
    'Gender',
    'Customer Type',
    'Type of Travel',
    'Class',
]
```

```python
# Variáveis categóricas ordinais (escala de 0 a 5)
categoricas_ordinais = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness'
]
```

```python
# Variáveis numéricas
numericas = [
    'Age',
    'Flight Distance',
    'Departure Delay in Minutes',
    'Arrival Delay in Minutes'
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numericas),
        ('cat', OneHotEncoder(), categoricas_nominais),
        ('ord', OrdinalEncoder(), categoricas_ordinais),
    ]
)
```

```python
# Aplicar Column Transformer
preprocessor.fit(df_airline_train)

# Exporta o modelo de pré-processamento
import joblib

joblib.dump(preprocessor, 'pkl/preprocessor_dataset_airline.pkl')
```
